# Ansatz

## Phase Modulation in Complex Exponential Form  
### General Case
Phase modulation operator $\Phi$ on a function $f(t)$    
$\Phi[f(t)] = e^{i\phi(t)} f(t)$  

## Decompose $\phi$ into Symmetric ($\phi_S$) and Antisymmetric ($\phi_A$) Parts  
$\phi(t) \equiv \phi_S(t) + \phi_A(t)$  

$\Phi[f(t)] = e^{i[\phi_S(t) + \phi_A(t)]} $  

## Fourier Transform  
$\widehat{\Phi[f(t)]} = \int{e^{i\omega t}e^{i\phi (t)}f(t)dt}$

$\widehat{\Phi[f(t)]} = \int{e^{i\omega t}e^{i[\phi_S(t) + \phi_A(t)]}f(t)dt}$

$\widehat{\Phi[f(t)]} = \int{e^{i[\omega t + \phi_S(t) + \phi_A(t)]}f(t)dt}$

## Special Case: $\phi(t) << 1$   
$\widehat{\Phi[f(t)]} \approx \int{e^{i\omega t}[1 + i\phi_S(t) + i\phi_A(t)]f(t)dt}$



$\widehat{\Phi[f(t)]} = \widehat{f(t)} + \int{e^{i\omega t}[i\phi_S(t) + i\phi_A(t)]f(t)dt}$


\begin{equation}
\widehat{\Phi[f(t)]} = \widehat{f(t)} + \int{[\cos(\omega t) + i\sin(\omega t)][i\phi_S(t) + i\phi_A(t)]f(t)dt} 
\tag{10} 
\end{equation}

## Further Specialization: $f(t) = 1$  
\begin{equation}
\widehat{\phi(t)} = \delta(0) + \int{[\cos(\omega t) + i\sin(\omega t)][i\phi_S(t) + i\phi_A(t)]dt}
\tag{20}
\end{equation}

$\widehat{\phi(t)} = \delta(0) + \int{\{i\cos(\omega t)[\phi_S(t) + \phi_A(t)] - \sin(\omega t)[\phi_S(t) + \phi_A(t)]\}dt}$

$\widehat{\phi(t)} = \delta(0) + \int{\{i\cos(\omega t)[\phi_S(t)] - \sin(\omega t)[\phi_A(t)]\}dt}$

$\widehat{\phi(t)} = \delta(0) + i\int{\cos(\omega t)\phi_S(t)}dt - \int{\sin(\omega t)\phi_A(t)dt}$

$\widehat{\phi(t)} = \delta(0) - \int{\sin(\omega t)\phi_A(t)dt} + i\int{\cos(\omega t)\phi_S(t)}dt$  

### Phase Transform 
$\theta \equiv \arctan\left[ \frac{\Im(\widehat{\phi(t)})}{\Re(\widehat{\phi(t)})}   \right]$


$\widehat{\phi(t)} = \delta(0) - \int{\sin(\omega t)\phi_A(t)dt} + i\int{\cos(\omega t)\phi_S(t)}dt$  

$\theta = \arctan\left[ \frac{\int{\cos(\omega t)\phi_S(t)dt}}{\delta(0)-\int{\sin(\omega t)\phi_A(t)dt}} \right]$  

$\theta(\omega) = \arctan\left[ \frac{\widehat{\phi_S(t)}}{\delta(0)-\widehat{\phi_A(t)}} \right]$  

### Power Transform
$I(\omega) = [\delta(0) - \widehat{\phi_A(t)} + i\widehat{\phi_S(t)}][\delta(0) - \widehat{\phi_A(t)} - i\widehat{\phi_S(t)}]$  

$I(\omega) = \delta(0)^2 - \delta(0)\widehat{\phi_A(t)} - i\delta(0)\widehat{\phi_S(t)} - \delta(0)\widehat{\phi_A(t)} + \widehat{\phi_A(t)}^2 +  i \widehat{\phi_A(t)}\widehat{\phi_S(t)} + i \delta(0)\widehat{\phi_S(t)} - i \widehat{\phi_A(t)}\widehat{\phi_S(t)} + \widehat{\phi_S(t)}^2$  

$I(\omega) = \delta(0)^2 - 2\delta(0)\widehat{\phi_A(t)} + \widehat{\phi_A(t)}\widehat{\phi_S(t)}+\widehat{\phi_A(t)}^2+\widehat{\phi_S(t)}^2$  

$I(\omega) = \delta(0)^2 + \widehat{\phi_A(t)}\widehat{\phi_S(t)}+\widehat{\phi_A(t)}^2+\widehat{\phi_S(t)}^2$  

## Generalize to All Real $f(t)$  

### Decompose $f$ to symmetric ($f_S$) and antisymmetric ($f_A$) Parts  
$f(t) \equiv f_S(t) + f_A(t)$  

Substitute into Eq. 1 
$\widehat{\Phi[f(t)]} = \widehat{f(t)} + \int{[\cos(\omega t) + i\sin(\omega t)][i\phi_S(t) + i\phi_A(t)][f_S(t) + f_A(t)]dt}$  

Expand:  
$\widehat{\Phi[f(t)]} = \widehat{f(t)} + \int{[\cos(\omega t) + i\sin(\omega t)][i\phi_S(t)f_S(t) + i\phi_S(t)f_A(t) + i\phi_A(t)f_S(t)+ i\phi_A(t)f_A(t)]dt}$  

### Define new variables for convenience:  
\begin{equation}
g_S(t) \equiv \phi_S(t)f_S(t) + \phi_A(t)f_A(t) 
\end{equation}  
\begin{equation}
g_A(t) \equiv \phi_S(t)f_A(t) + \phi_A(t)f_S(t)
\end{equation}

Substitute:  
$\widehat{\Phi[f(t)]} = \widehat{f(t)} + \int{[\cos(\omega t) + i\sin(\omega t)][ig_S(t) + ig_A(t)]dt}$  

This equation is isomorphic to Eq. 20, with $\phi_S \leftrightarrow g_S$ and $\phi_A \leftrightarrow g_A$.  

$\therefore$  
$\widehat{\Phi[f(t)]} = \widehat{f(t)} + \int{[\cos(\omega t) + i\sin(\omega t)][ig_S(t) + ig_A(t)]dt}$  

This equation is isomorphic to Eq. 20, with $\phi_S \leftrightarrow g_S$ and $\phi_A \leftrightarrow g_A$.  

### Phase Transform (of Small Phase Modulation on Real Function)  
\begin{equation}
\theta(\omega) = \arctan\left[ \frac{\widehat{g_S(t)}}{\delta(0)-\widehat{g_A(t)}} \right]
\end{equation}

# Phase modulation on a real function yields an antisymmetric phase transform  

### Power Transform (of Small Phase Modulation on Real Function)   
\begin{equation}
I(\omega) = \delta(0)^2 + \widehat{g_A(t)}\widehat{g_S(t)}+\widehat{g_A(t)}^2+\widehat{g_S(t)}^2
\end{equation}

# Compared to amplitude modulation, small phase modulation adds a $\delta(0)^2$ term to the power transform.  

```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
n_pts = 1024
x = np.linspace(0, 1, n_pts)
```

```python
# Symmetric Part
A = 1e-3
# A = 0.0
w = 0.005
sym = A*np.exp(-(x/w)**2)
```

```python
plt.title('SYMMETRIC PART')
plt.plot(x, sym, 'k')
plt.grid(True, 'both')
```

```python
# Antisymmetric Part
B = 1e-3
win_w = 0.01
window = np.exp(-x**2/win_w)
# B = 0.0
antisym = B*np.sin(2*np.pi*4*x)*window
```

```python
plt.title('ANTI-SYMMETRIC PART')
plt.plot(x, antisym, 'k')
plt.grid(True)
```

```python
# Phase Modulation Function
f = sym + antisym
```

```python
plt.title('TOTAL PHASE MODULATION FUNCTION')
plt.plot(x, f, 'k')
plt.grid(True)
```

```python
# Modulated Wave
wv = np.exp(1j*f)
# wv = f
# wv = np.exp(1j*2*np.pi*50*x)*np.exp(1j*f)
```

```python
plt.title('MODULATED WAVE AMPLITUDE')
plt.plot(x, np.absolute(wv), 'r')
plt.grid(True)
```

```python
plt.title('MODULATED WAVE PHASE')
plt.plot(x, np.angle(wv), 'r')
plt.grid(True)
```

```python
# Fourier Transform  
wv_fft = np.fft.fft(wv)
delta_x = x[1]-x[0]
p = np.fft.fftfreq(n_pts, delta_x)
# p = np.fft.fftshift(np.fft.fftfreq(n_pts, delta_x))
amplitude = np.absolute(wv_fft)
phase = np.angle(wv_fft)
# phase = np.arctan2(np.imag(wv_fft), np.real(wv_fft))
```

```python
plt.title('FFT AMPLITUDE')
plt.plot(p, amplitude, '.')
# plt.xlim(90, 110)
plt.grid(True)
```

```python
plt.title('FFT PHASE')
plt.plot(p, phase/np.pi, '.')
# plt.xlim(90, 110)
plt.ylabel('phase [pi]')
plt.grid(True)
```

```python
plt.title('FFT REAL')
plt.plot(p, np.real(wv_fft), '.')
# plt.xlim(-1, 1)
plt.grid(True)
```

```python
plt.title('FFT IMAGINARY')
plt.plot(p, np.imag(wv_fft), '.')
# plt.xlim(-10, 10)
plt.grid(True)
```
