# Phase Modulation and Its Effect on Fourier Transform  

## Phase Modulation in Complex Exponential Form  
### General Case
Phase modulation operator $\Phi$ on a function $f(t)$    
$\Phi[f(t)] = e^{i\phi(t)} f(t)$  

## Decompose $\phi$ into Symmetric ($\phi_S$) and Antisymmetric ($\phi_A$) Parts  
$\phi(t) \equiv \phi_S(t) + \phi_A(t)$  

$\Phi[f(t)] = e^{i[\phi_S(t) + \phi_A(t)]} f(t)$  

## Fourier Transform  

$\hat{\Phi}(\omega) = \int{e^{i\omega t}e^{i\phi (t)}f(t)dt}$

Notational simplifications:  
* $\hat{\Phi}(\omega) \equiv \widehat{\Phi[f(t)]}(\omega)$, where the caret denotes Fourier transform.    
* Integration limits are $\infty$ and $-\infty$, unless otherwise noted.  


$\hat{\Phi}(\omega) = \int{e^{i\omega t}e^{i[\phi_S(t) + \phi_A(t)]}f(t)dt}$

$\hat{\Phi}(\omega) = \int{e^{i[\omega t + \phi_S(t) + \phi_A(t)]}f(t)dt}$

## Special Case: $\phi(t) << 1$   
$\hat{\Phi}(\omega) \approx \int{e^{i\omega t}[1 + i\phi_S(t) + i\phi_A(t)]f(t)dt}$



$\hat{\Phi}(\omega) \approx \hat{f}(\omega) + \int{e^{i\omega t}[i\phi_S(t) + i\phi_A(t)]f(t)dt}$


\begin{equation}
\hat{\Phi}(\omega) \approx \hat{f}(\omega) + \int{[\cos(\omega t) + i\sin(\omega t)][i\phi_S(t) + i\phi_A(t)]f(t)dt} 
\tag{10} 
\end{equation}

## Further Specialization: $f(t) = 1$  
\begin{equation}
\hat{\Phi}(\omega) \approx \delta(0) + \int{[\cos(\omega t) + i\sin(\omega t)][i\phi_S(t) + i\phi_A(t)]dt}
\tag{20}
\end{equation}

$\hat{\Phi}(\omega) \approx \delta(0) + \int{\{i\cos(\omega t)[\phi_S(t) + \phi_A(t)] - \sin(\omega t)[\phi_S(t) + \phi_A(t)]\}dt}$

$\hat{\Phi}(\omega) \approx \delta(0) + \int{\{i\cos(\omega t)[\phi_S(t)] - \sin(\omega t)[\phi_A(t)]\}dt}$

$\hat{\Phi}(\omega) \approx \delta(0) + i\int{\cos(\omega t)\phi_S(t)}dt - \int{\sin(\omega t)\phi_A(t)dt}$

$\hat{\Phi}(\omega) \approx \delta(0) - \int{\sin(\omega t)\phi_A(t)dt} + i\int{\cos(\omega t)\phi_S(t)}dt$  

### Phase Transform 
$\theta(\omega) \equiv \arctan\left[ \frac{\Im(\hat{\Phi}(\omega))}{\Re(\hat{\Phi}(\omega))}   \right]$


$\hat{\Phi}(\omega) \approx \delta(0) - \int{\sin(\omega t)\phi_A(t)dt} + i\int{\cos(\omega t)\phi_S(t)}dt$  

$\theta(\omega) \approx \arctan\left[ \frac{\int{\cos(\omega t)\phi_S(t)dt}}{\delta(0)-\int{\sin(\omega t)\phi_A(t)dt}} \right]$  

$\theta(\omega) \approx \arctan\left[ \frac{\hat{\Phi}_S(\omega)}{\delta(0)-\hat{\Phi}_A(\omega)} \right]$  

### Power Transform
$I(\omega) \approx [\delta(0) - \hat{\Phi}_A(\omega) + i\hat{\Phi}_S(\omega)][\delta(0) - \hat{\Phi}_A(\omega) - i\hat{\Phi}_S(\omega)]$  

$I(\omega) \approx \delta(0)^2 - \delta(0)\hat{\Phi}_A(\omega) - i\delta(0)\hat{\Phi}_S(\omega) - \delta(0)\hat{\Phi}_A(\omega) + \hat{\Phi}_A(\omega)^2 +  i \hat{\Phi}_A(\omega)\hat{\Phi}_S(\omega) + i \delta(0)\hat{\Phi}_S(\omega) - i \hat{\Phi}_A(\omega)\hat{\Phi}_S(\omega) + \hat{\Phi}_S(\omega)^2$  

$I(\omega) \approx \delta(0)^2 - 2\delta(0)\hat{\Phi}_A(\omega) + \hat{\Phi}_A(\omega)^2+\hat{\Phi}_S(\omega)^2$  

$I(\omega) \approx \delta(0)^2 + \hat{\Phi}_A(\omega)^2+\hat{\Phi}_S(\omega)^2$  

## Generalize to All Real $f(t)$  

### Decompose $f$ to symmetric ($f_S$) and antisymmetric ($f_A$) Parts  
$f(t) \equiv f_S(t) + f_A(t)$  

Substitute into Eq. 10   

$\hat{\Phi}(\omega) \approx \hat{f}(\omega) + \int{[\cos(\omega t) + i\sin(\omega t)][i\phi_S(t) + i\phi_A(t)][f_S(t) + f_A(t)]dt}$  

Expand:  
$\hat{\Phi}(\omega) \approx \hat{f}(\omega) + \int{[\cos(\omega t) + i\sin(\omega t)][i\phi_S(t)f_S(t) + i\phi_S(t)f_A(t) + i\phi_A(t)f_S(t)+ i\phi_A(t)f_A(t)]dt}$  

### Define new variables for convenience:  
\begin{equation}
g_S(t) \equiv \phi_S(t)f_S(t) + \phi_A(t)f_A(t) 
\end{equation}  
\begin{equation}
g_A(t) \equiv \phi_S(t)f_A(t) + \phi_A(t)f_S(t)
\end{equation}

$g_S$ is symmetric, and $g_A$ is anti-symmetric.  

$\therefore$  

\begin{equation}  
\hat{g}_S = \hat{\phi}_S(\omega)\star\hat{f}_S(\omega) + \hat{\phi}_A(\omega)\star\hat{f}_A(\omega)
\tag{30}
\end{equation}

\begin{equation}  
\hat{g}_A = \hat{\phi}_S(\omega)\star\hat{f}_A(\omega) + \hat{\phi}_A(\omega)\star\hat{f}_S(\omega)
\tag{40}
\end{equation}

$\star$ denotes convolution.  


Substitute:  
$\hat{\Phi}(\omega) \approx \hat{f}(\omega) + \int{[\cos(\omega t) + i\sin(\omega t)][ig_S(t) + ig_A(t)]dt}$  

This equation is isomorphic with Eq. 20, with $\delta(0) \leftrightarrow \hat{f}(\omega)$, $\phi_S \leftrightarrow g_S$ and $\phi_A \leftrightarrow g_A$.    

$\therefore$  

$\hat{\Phi}(\omega) \approx  \hat{f}(\omega) - \int{\sin(\omega t)g_A(t)dt} + i\int{\cos(\omega t)g_S(t)}dt$  

$\hat{\Phi}(\omega) \approx  \int{\cos(\omega t)f_S(t)dt} + i\int{\sin(\omega t)f_A(t)}dt- \int{\sin(\omega t)g_A(t)dt} + i\int{\cos(\omega t)g_S(t)}dt$  



### Phase Transform (of Small Phase Modulation on Real Function)  
\begin{equation}
\theta(\omega) \approx \arctan\left[ \frac{\hat{f}_A(\omega) + \hat{g}_S(\omega)}{\hat{f}_S(\omega)-\hat{g}_A(\omega)} \right]
\tag{50}
\end{equation}  

Substitute in Equations 30 and 40:  

\begin{equation}
\theta(\omega) \approx \arctan\left[ \frac{\hat{f}_A(\omega) + \hat{\phi}_S\omega)\star\hat{f}_S(\omega) + \hat{\phi}_A(\omega)\star\hat{f}_A(\omega)}{\hat{f}_S(\omega)-\hat{\phi}_S(\omega)\star\hat{f}_A(\omega) - \hat{\phi}_A(\omega)\star\hat{f}_S(\omega)} \right]
\tag{60}
\end{equation}

### Power Transform (of Small Phase Modulation on Real Function)   
\begin{equation}
I(\omega) \approx \hat{f}(\omega)^2 + \hat{g}_A(\omega)^2+\hat{g}_S(\omega)^2
\tag{70}
\end{equation}

## Conclusions   

In the case of a real amplitude function combined with a real phase modulation, one can obtain a Fourier conjugate with *arbitrary symmetry* in phase (*cf.* Equation 60).  

In particular, obtaining a Fourier conjugate with a *symmetric* (non-zero) phase function requires one of the following scenarios:  
1. The amplitude function ($f$) and phase modulation ($\phi$) are both *symmetric*.  
    * In this case, $\theta(\omega) \approx \arctan\left[ \frac{\hat{\phi}_S(\omega)\star\hat{f}_S(\omega) }{\hat{f}_S(\omega)} \right]$
2. The amplitude function ($f$) is *antisymmetric*, but the phase function ($\phi$) is *symmetric*.
    * In this case, $\theta(\omega) \approx \arctan\left[ \frac{\hat{f}_A(\omega)  }{\hat{\phi}_S(\omega)\star\hat{f}_A(\omega) } \right]$
