```python
import time
import numpy as np
import matplotlib.pyplot as plt
import hcipy_local.hcipy as hci
import sleec_input as sli
```

```python
coro_1x = sli.ini_corngrph(sli.default_args())
wf_1x = hci.Wavefront(coro_1x.pupil_field, coro_1x.wvl)
```

```python
t0 = time.time()
prop_1x = coro_1x(wf_1x)
print("time elapsed = {}".format(time.time()-t0))
ref_intnsty_1x = prop_1x['on_fpm'].intensity.max()
```

```python
coro_4x = sli.ini_corngrph(sli.input_args())
wf_4x = hci.Wavefront(coro_4x.pupil_field, coro_4x.wvl)
```

```python
t0 = time.time()
prop_4x = coro_4x(wf_4x)
print("time elapsed = {}".format(time.time()-t0))
ref_intnsty_4x = prop_4x['on_fpm'].intensity.max()
```

```python
plt.figure(figsize=(20, 10))
plt.subplot(121)
hci.imshow_field(prop_1x['pupil'].intensity, cmap='inferno')
plt.colorbar()
plt.subplot(122)
hci.imshow_field(prop_4x['pupil'].intensity, cmap='inferno')
plt.colorbar()
plt.show()
```
