```python
import numpy as np
import matplotlib.pyplot as plt
import hcipy as hci
from hcipy import make_disk_harmonic_basis as mdhb
```

```python
grid = hci.make_pupil_grid(480, 1)
```

```python
basis = mdhb(grid, 9, bc='dirichlet')
```

```python
plt.figure(figsize=(16,9))
plt.subplot(331)
hci.imshow_field(basis[0])
plt.colorbar()
plt.subplot(332)
hci.imshow_field(basis[1])
plt.colorbar()
plt.subplot(333)
hci.imshow_field(basis[2])
plt.subplot(334)
hci.imshow_field(basis[3])
plt.subplot(335)
hci.imshow_field(basis[4])
plt.subplot(336)
hci.imshow_field(basis[5])
plt.subplot(337)
hci.imshow_field(basis[6])
plt.subplot(338)
hci.imshow_field(basis[7])
plt.subplot(339)
hci.imshow_field(basis[8])
```

```python
rng = np.random.default_rng()
coefs = rng.standard_normal(9)
print(coefs)
combo = basis.linear_combination(coefs)
hci.imshow_field(combo)
```

```python
dim = int(np.sqrt(combo.shape[0]))
print(dim)
```

```python
plt.plot(combo.reshape(dim,dim)[dim//2, :])
plt.grid(True, 'both')
```
